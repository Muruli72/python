```python
import pandas as pd

data = {
    "Item": ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"],
    "Cost": [75000, 500, 1500, 12000, 2000],
    "Quantity": [1, 2, 1, 1, 2],
    "Date": ["2026-05-01", "2026-05-02", "2026-05-01", "2026-05-03", "2026-05-02"]
}

df = pd.DataFrame(data)

# Save to CSV
df.to_csv("items.csv", index=False)

print("CSV file created!")
```

```python
import pandas as pd

# Read CSV
df = pd.read_csv("items.csv")

# Convert Date column to datetime (important for filtering)
df["Date"] = pd.to_datetime(df["Date"])

# Ask user for date
user_date = input("Enter date (YYYY-MM-DD): ")

# Filter rows
filtered = df[df["Date"] == user_date]

print(filtered)
```
