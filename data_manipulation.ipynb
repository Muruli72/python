```python
def intersect( nums1, nums2):
        a = []
        for i in nums1:
            for j in nums2:
                if i==j :
                   a.append(i)
                   nums1.remove(i)
                   nums2.remove(j)
           
        return a
nums1 = [1,2,2,1]
nums2 = [2,2]
print(intersect(nums1, nums2))
```
