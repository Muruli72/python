```python
import cv2
image=cv2.imread("luffy.jpg")
```

```python
print(image)
```

```python
import matplotlib.pyplot as plt

plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
print(image.size)
print(image.shape)
```

```python
#image change BGR to RGB
img=cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
plt.imshow(img)
plt.axis("off")
```

```python
#image change BGR to RGB
img=cv2.cvtColor(image,cv2.COLOR_RGB2BGR)
plt.imshow(img)
plt.axis("off")
```

```python
gray_image=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
plt.imshow(gray_image,cmap="gray")
plt.axis("off")
```

```python
H,W,c=image.shape
```

```python
print("Resolution:",W,"*",H,"Pixels")
```

```python
cv2.imwrite("luffy1.jpeg",image)
```

```python
#resize
resized_image=cv2.resize(gray_image,(400,300))
```

```python
plt.imshow(resized_image, cmap="gray")
print("Reszed Image deatils")
height, width = resized_image.shape
print("Size:", resized_image.size)
print("Shape:", resized_image.shape)
print("Width:", width, "Height:", height)

plt.axis("off")
plt.show()
```

```python
rotate_90=cv2.rotate(image,cv2.ROTATE_90_CLOCKWISE)
plt.imshow(rotate_90,cmap="gray")
plt.axis("off")
```

```python
croped_image=resized_image[100:400,100:300]
plt.imshow(croped_image,cmap='gray')
plt.axis("off")
```
