```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


image = cv2.imread("input.jpeg")

if image is None:
    print("Error: Image not found!")
    exit()


image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)



laplacian = cv2.Laplacian(
    image_rgb,
    cv2.CV_64F
)

laplacian = cv2.convertScaleAbs(laplacian)

laplacian_sharp = cv2.addWeighted(
    image_rgb,
    1.0,
    laplacian,
    1.0,
    0
)



blurred = cv2.GaussianBlur(
    image_rgb,
    (5, 5),
    0
)

unsharp_mask = cv2.subtract(
    image_rgb,
    blurred
)

unsharp = cv2.addWeighted(
    image_rgb,
    1.0,
    unsharp_mask,
    1.0,
    0
)



A = 2.0

high_boost = cv2.addWeighted(
    image_rgb,
    A,
    blurred,
    -1.0,
    0
)



plt.figure(figsize=(12, 8))

plt.subplot(3, 3, 1)
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(3, 3, 2)
plt.imshow(laplacian_sharp)
plt.title("Laplacian Sharpening")
plt.axis("off")
plt.subplot(3, 3, 3)
plt.imshow(unsharp_mask)
plt.title("Unsharp Masking")
plt.axis("off")
plt.subplot(3, 3, 4)
plt.imshow(unsharp)
plt.title("Unsharp Masking")
plt.axis("off")

plt.subplot(3, 3, 5)
plt.imshow(high_boost)
plt.title("High-Boost Filtering")
plt.axis("off")

plt.tight_layout()
plt.show()
```
