```python
import cv2
import matplotlib.pyplot as plt
```

```python
image=cv2.imread("luffy.jpg")
```

```python
plt.imshow(image)
plt.axis("off")
```

```python
image_rgb=cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(image_rgb)
plt.axis("off")
plt.show()
```

```python
h,w,c=image.shape
```

```python
print("Height:",h)
print("width:",w)
print("Channle",c)
```

```python
print("Resolution :", w, "x", h, "pixels")
```

```python
cv2.imwrite("l.jpg",image_rgb)
```

```python
resize_img=cv2.resize(image,(400,300))
```

```python

plt.imshow(cv2.cvtColor(resize_img, cv2.COLOR_BGR2RGB))
plt.title("Resized Image")
plt.axis("off")
plt.show()
```

```python
cropped_img=image[100:400,100:500]
```

```python

plt.imshow(cv2.cvtColor(cropped_img, cv2.COLOR_BGR2RGB))
plt.title("Cropped Image")
plt.axis("off")
plt.show()
```

```python
rotated_image=cv2.rotate(image_rgb,cv2.ROTATE_90_COUNTERCLOCKWISE)
```

```python
plt.imshow(cv2.cvtColor(rotated_image, cv2.COLOR_BGR2RGB))
plt.title("Rotated Image")
plt.axis("off")
plt.show()
```

```python
# 9. Convert to Grayscale
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2.imwrite("grayscale_image.jpg", gray_image)

plt.imshow(gray_image, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

```python
print("\nPixel Intensity Information")
print("---------------------------")
print("Minimum intensity :", gray_image.min())
print("Maximum intensity :", gray_image.max())
print("Mean intensity    :", gray_image.mean())
```

```python
x = 100
y = 100

if y < gray_image.shape[0] and x < gray_image.shape[1]:
    print("Pixel intensity at (100,100) :", gray_image[y, x])

# 12. Image Data Type
print("\nAdditional Image Information")
print("----------------------------")
print("Image Data Type :", image.dtype)
print("Grayscale Shape :", gray_image.shape)
```
