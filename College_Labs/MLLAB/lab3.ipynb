```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error
```

```python
df=pd.read_csv('housing.csv')
```

```python
df.head()
```

```python
df.dtypes
```

```python
df.describe()
```

```python
print(df.columns.to_list())
```

```python
target='median_house_value'
```

```python
x=df.drop(columns=target)
```

```python
y=df[target]
```

```python
print("Before converstion")
x.head()
x=pd.get_dummies(x,drop_first=True)
x.head()
```

```python
x.head()
```

```python
print(df.isnull().sum())
```

```python
from sklearn.impute import SimpleImputer
```

```python
imputer=SimpleImputer(strategy='median')
```

```python
x_imputed_array=imputer.fit_transform(x)
```

```python
x.head()
```

```python
print(x.isnull().sum())
```

```python
x=pd.DataFrame(x_imputed_array,columns=x.columns,index=x.index)
```

```python
X_train, X_test, y_train, y_test = train_test_split(   x,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
linear= LinearRegression()
```

```python
linear.fit(X_train,y_train)
y_pred_linear=linear.predict(X_test)
print("Actual values")
print(y_test.values)
print("predicted")
print(y_pred_linear)
```

```python
mae_linear=mean_absolute_error(y_test,y_pred_linear)
mse_linear=mean_squared_error(y_test,y_pred_linear)
rmse_linear=np.sqrt(mse_linear)
r2_linear=r2_score(y_test,y_pred_linear)
```

```python
print("MAE:",round(mae_linear,2))
print("MSE",round(mse_linear,2))
print("RMSE:",round(rmse_linear,2))
print("R@ SCore:",round(r2_linear),2)
```

```python
poly=PolynomialFeatures(degree=2,include_bias=False)
```

```python
x_train_poly=poly.fit_transform(X_train)
x_test_poly=poly.transform(X_test)
poly_model=LinearRegression()
poly_model.fit(x_train_poly,y_train)
y_pred_poly=poly_model.predict(x_test_poly)
print("Actual values")
print(y_test.values)
print("predicted")
print(y_pred_poly)
```

```python
mae_poly=mean_absolute_error(y_test,y_pred_poly)
mse_poly=mean_squared_error(y_test,y_pred_poly)
rmse_poly=np.sqrt(mse_poly)
r2_poly=r2_score(y_test,y_pred_poly)
```

```python
print("MAE:",round(mae_poly,2))
print("MSE",round(mse_poly,2))
print("RMSE:",round(rmse_poly,2))
print("R@ SCore:",round(r2_poly,2))
```
