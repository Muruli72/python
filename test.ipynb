```python
###modeling a bank account
class BankAccount:
    def __init__(self,account_number,account_holder,balance=0):
        self.account_number=account_number
        self.account_holder=account_holder
        self.balance=balance

    def deposit(self,amount):
        if amount>0:
            self.balance+=amount
            print(f"Deposited {amount}. New balance: {self.balance}")
        else:
            print("Invalid deposit amount.")

    def withdraw(self,amount):
        if amount>0 and amount<=self.balance:
            self.balance-=amount
            print(f"Withdrew {amount}. New balance: {self.balance}")
        else:
            print("Invalid withdrawal amount or insufficient funds.")
```

```python
bank_account=BankAccount(account_number="123456789",account_holder="John Doe")
bank_account.deposit(1000)
bank_account.withdraw(500)
bank_account.withdraw(600)
```

```python
from abc import ABC, abstractmethod
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
class Circle(Shape):
    def __init__(self,radius):
        self.radius=radius

    def area(self):
        return 3.14*self.radius**2
circle=Circle(radius=5)
print(circle.area())    
```

```python
def continueSquence(nums):
    count=0
    nums=set(nums)
    for num in nums:
        if num - 1 not in nums:
            curr_count=1
            while num+1 in nums:
                num+=1
                curr_count+=1
            count=max(count,curr_count)
    return count
```

```python
s=[100, 4, 200, 1, 3, 2,5,988,7,6]
print(continueSquence(s))
```

```python
x=121
rem=0
def is_palindrome(x):
    original_x=x
    rem=0
    while x>0:
        num=x%10
        rem=rem*10+num
        x=x//10
    return rem==original_x
print(is_palindrome(x))
```

```python
def TwoSum(nums,target):
    for i in range(len(nums)):
        for j in range(i+1,len(nums)):
            if nums[i]+nums[j]==target:
                return [i,j]
nums=[2,7,11,15]
target=9
print(TwoSum(nums,target))
```

```python
nums=[11,2,3,18,25,4,45,1]
def BubbleSort(nums):
    n=len(nums)
    for i in range(n):
        flag=False
        for j in range(0,n-i-1):
            if nums[j]>nums[j+1]:
                nums[j],nums[j+1]=nums[j+1],nums[j]
                flag=True
        if  flag==False:
             break
    return nums
print(BubbleSort(nums))
```
