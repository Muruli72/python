```python
nums = [1, 3, 4, 2, 2,4]

seen = set()

for num in nums:
    if num in seen:
        print("Duplicate:", num)
    seen.add(num)
    print("Seen so far:", seen)
print("Final seen set:", seen)

```

```python
def present_or_not(nums,t):
    for i in range(len(nums)):
        for j in range(len(nums)):
            if nums[i]+nums[j]==t:
                return True
            else:
                False

```

```python

nums=[2,5,7,8,19];t=1
print(present_or_not(nums,t))
```

```python
import pandas as pd

person = pd.DataFrame({
    'id': [1, 2, 3, 4],
    'email': [
        'john@example.com',
        'bob@example.com',
        'john@example.com',
        'alice@example.com'
    ]
})

print(person)
import pandas as pd

def delete_duplicate_emails(person: pd.DataFrame) -> None:
    person.drop_duplicates(
        subset='email',
        keep='first',
        inplace=True
    )
delete_duplicate_emails(person)

print("\nAfter Removing Duplicates:")
print(person)
```
