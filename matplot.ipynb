```python
x=[1,2,3,4,5]
y=[1,4,5,6,8]

plt.plot(x,y)
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title("Basic Line plot")
plt.show()

```

```python
x=[1,2,3,4,5]
y=[1,4,5,6,8]

plt.plot(x,y,color='Blue',linestyle='-.',marker='o',linewidth=5)
```

```python
## Multiple Plots
##sample data
x=[1,2,3,4,5]
y1=[2,3,5,7,9]
plt.figure(figsize=(9,4))
plt.subplot(2,2,1)
plt.plot(x,y1,color='Yellow')
plt.title('plot 1')

plt.subplot(2,2,2)
plt.plot(y1,x,color='orange')
plt.title('plot 2')
```

```python
import sqlite3
```

```python
connection=sqlite3.connect('example.db')
print(connection)
```

```python
cursor=connection.cursor()
```

```python
cursor.execute("CREATE TABLE IF NOT EXISTS students (id INTEGER, name TEXT)")

# Insert data
cursor.execute("INSERT INTO students VALUES (1, 'Harish')")

# Save changes
connection.commit()

# Fetch data
cursor.execute("SELECT * FROM students")
rows = cursor.fetchall()

print(rows)

# Close connection
connection.close()
```
